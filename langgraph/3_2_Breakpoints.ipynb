# Точки останова (Breakpoints)

## Обзор

Для `человека в цикле` (human-in-the-loop) мы часто хотим видеть выходные данные нашего графа (graph outputs) по мере его выполнения.

Мы заложили основы для этого с помощью потоковой передачи (streaming).

## Цели

Теперь давайте обсудим мотивации для использования `человека в цикле` (human-in-the-loop):

(1) `Утверждение` (Approval) - Мы можем прервать работу нашего агента (agent), показать состояние (state) пользователю и позволить пользователю подтвердить действие.

(2) `Отладка` (Debugging) - Мы можем перемотать граф (rewind the graph), чтобы воспроизвести или избежать проблем.

(3) `Редактирование` (Editing) - Вы можете изменить состояние (state).

LangGraph предлагает несколько способов получения или обновления состояния агента (agent state) для поддержки различных рабочих процессов с участием `человека в цикле` (human-in-the-loop workflows).

Сначала мы представим [точки останова](https://langchain-ai.github.io/langgraph/how-tos/human_in_the_loop/breakpoints/#simple-usage) (breakpoints), которые предоставляют простой способ остановить граф (graph) на определенных шагах.

Мы покажем, как это позволяет пользователю `утверждать` (approval) действия.

## Точки останова (Breakpoints) для утверждения человеком (human approval)

Давайте снова рассмотрим простого агента (agent), с которым мы работали в Модуле 1.

Предположим, что нас беспокоит использование инструментов (tool use): мы хотим утверждать использование агентом любого из его инструментов.

Все, что нам нужно сделать, это просто скомпилировать граф (graph) с параметром `interrupt_before=["tools"]`, где `tools` — это узел (node) наших инструментов.

Это означает, что выполнение будет прервано перед узлом `tools`, который выполняет вызов инструмента (tool call).
